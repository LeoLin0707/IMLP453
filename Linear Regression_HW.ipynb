## 用訓練資料來 fit 函數
1. 只用訓練資料集的資料進行linear regression演算法<br>
2. 並計算出訓練階段的MSE
3. 畫出目標點(藍色)與預測點(紅色)的對應圖
![](images/LR_train.PNG)

```python
X_t = x_train.reshape(-1,1)
x_line.fit(X_t,y_train)
Y_pre = x_line.predict(X_t)
MSE_t = np.sum((Y_pre-y_train)**2) / len(y_train)
W = x_line.coef_
B = x_line.intercept_
print(type(W[0]))
print(str(W[0])+" "+str(round(B,2)))
```

```python
title='f(x)='+str(round(W[0],2))+'x+'+str(round(B,2))+',mse:'+str(round(MSE_t,2))
plt.title(title)
plt.scatter(x_train,y_train,c='blue',label='Ground Truth')
plt.scatter(x_train,Y_pre,c='red',label='Predict')
plt.legend()
plt.show()
```
